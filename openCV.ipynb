```python
import cv2
```

```python
image = cv2.imread("pic1.jpeg")

if image is not None:
    cv2.imshow("Image", image)
    cv2.waitKey(0) #waitkey is used to wait a key press to close the window and 0 is used to wait indefinitely
    cv2.destroyAllWindows()
else:
    print("Image not found...")
```

```python
grey_image = cv2.cvtColor(image, cv2.COLOR_BGR2GRAY)  #to change the color of image
cv2.imshow('Gray Image', grey_image)
cv2.waitKey(0)
```

```python
blurred_image = cv2.GaussianBlur(image, (15,15), 0) #(15,15) is the kernel size and kernel size means the size of the filter
cv2.imshow("Blurred Image", blurred_image)
cv2.waitKey(0)
```

```python
edges = cv2.Canny(grey_image, 100, 200) #100 is the lowerthreshold and 200 is the upper threshold, threshold is the minimum value of the gradient
cv2.imshow("Edges", edges)
cv2.waitKey(0)
```

```python
image = cv2.imread("pic1.jpeg")
print(image.shape)  #660 is the height, 1268 is the width and 3 is the no of channels, channels are the no of shape

#resize image
resizedImage = cv2.resize(image, (300,300))     #(300,300) is the new size of the image

#cropped image
cropeedImage = image[0:200, 200:500]     # [0:200, 200:500] is the height and width of the image

cv2.imshow('Image',image)
cv2.imshow('Resized image', resizedImage)
cv2.imshow('Cropped image', cropeedImage)
cv2.waitKey(0)
```

```python
drawing = image.copy()
cv2.rectangle(drawing, (50,50), (200,200), (0,200,0), 2) #(50,50) is the staring point, (200,200) is the ending point, (0,200,0) is the color of the rectangle and 2 is the thickness of the rectangle
cv2.circle(drawing, (300,300), 50, (0,0,255), 2) #(300,300) is the center of the circle50 is the radius of the circle, (0,0,255) is the color of the circle and 2 is the thickness of the circle
cv2.imshow("Drawing", drawing)
cv2.waitKey(0)
```
