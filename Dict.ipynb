```python
student = {
    'name' : 'John',
    'class' : '12th',
    'roll_no' : 12,
    'age' : 17
}
print(student)
```

```python
temp ={
    'lucknow' : [25,30,35],
    'delhi': [30,35,40],
    'chennai' : [24,23,22]
}
print(temp)
```

```python
#to merge two dictionary
student.update(temp)
print(student)
```

```python
#to iterate a dictionary using for loop
for k, v in student.items():
    print(f'{k} : {v}')
```

```python
sales = {'India':100, 'China':500, 'USA':800, 'Japan':700}
print(sales)
```

```python
#sort by key
sales_keys = sorted(sales)
print(sales_keys)
```

```python
#sort by values
sales = {k: v for k, v in sorted(sales.items(),key=lambda x: x[1])}
print(sales)
```

```python
#delete an element
del sales['India']
print(sales)
```
